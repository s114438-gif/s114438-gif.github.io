## 傳送門裝置 `TeleportPortal`

接下來進入第六章前,先認識前輩留下的傳送門裝置。

下方的 cell 是傳送門的完整實作,**不要修改**,直接執行該 cell 讓 Python 記住這個類別即可。

### TeleportPortal 使用手冊

| 方法 | 用途 | 回傳值 |
|---|---|---|
| `portal = TeleportPortal(initial_energy)` | 建立傳送門實例,傳入起始能量 | 物件本身 |
| `portal.charge(inventory)` | 投入 inventory 中所有物資充能 | 目前總能量(int) |
| `portal.set_destination(place)` | 設定目的地字串 | 無 |
| `portal.check_weather()` | 重擲天氣,印出並回傳當前天氣 | `'rainy'` / `'cloudy'` / `'sunny'` |
| `portal.get_energy()` | 查詢目前總能量 | 目前總能量(int) |
| `portal.check_ready()` | 檢查是否可啟動 | True / False |
| `portal.fire()` | 啟動傳送門 | 結果字串(`"SUCCESS: ..."` 或 `"FAILED: ..."`) |

**啟動條件(由 `check_ready` 內部判斷):**
- 目的地必須已設定
- `rainy` 永遠不可啟動
- `cloudy` 需要 energy >= 80
- `sunny` 需要 energy >= 50

```python
# ============================================================
# 傳送門裝置(老師預寫,請勿修改,直接執行此 cell 即可)
# ============================================================

class TeleportPortal:
    """前輩留下的傳送門裝置。學員不需要理解內部實作,只要照手冊呼叫即可。"""

    def __init__(self, initial_energy: int):
        self._energy = initial_energy
        self._destination = None
        self._weather = 'rainy'
        self._ready = False
        self._charged = False

    def charge(self, inventory: dict) -> int:
        """投入背包中的物資,自動轉換為能量。回傳目前總能量。"""
        if self._charged:
            return self._energy
        gained = 0
        for item, count in inventory.items():
            if count <= 0:
                continue
            char_sum = sum(ord(c) for c in item)
            length_factor = abs(len(item) - 4) + 1
            rarity = max(2, char_sum % 13)
            item_base = (rarity + length_factor) * 2
            count_boost = math.sqrt(count) + 0.5 * count
            e = int(item_base * count_boost) + 12
            gained += e
        self._energy += gained
        self._charged = True
        return self._energy

    def set_destination(self, place: str) -> None:
        """設定傳送目的地。"""
        self._destination = place

    def check_weather(self) -> str:
        """重新擲骰決定當前天氣,印出並回傳天氣字串。
        rainy: 3/6, cloudy: 1/6, sunny: 2/6"""
        self._weather = random.choices(
            ['rainy', 'cloudy', 'sunny'],
            weights=[3, 1, 2]
        )[0]
        print(f"[氣象] 當前天氣:{self._weather}")
        if self._weather == 'rainy':
            print("爛台北,下雨天拉完了! 戳手手")
        return self._weather

    def get_energy(self) -> int:
        return self._energy

    def check_ready(self) -> bool:
        """檢查是否可啟動。"""
        if self._destination is None:
            self._ready = False
        elif self._weather == 'rainy':
            self._ready = False
        elif self._weather == 'cloudy':
            self._ready = self._energy >= 80
        elif self._weather == 'sunny':
            self._ready = self._energy >= 50
        return self._ready

    def fire(self) -> str:
        """啟動傳送門,回傳結果字串。"""
        if not self._ready:
            return "FAILED: portal not ready."
        return f"SUCCESS: teleporting to {self._destination}..."
```
